```python
import pandas as pd
df = pd.read_csv("StudentsPerformance.csv")
```

#### Question: Does parental level of education impact the overall score of students?

```python
df["average score"] = (df["math score"] + df["reading score"] + df["writing score"]) / 3
print(df.groupby("parental level of education")[["average score"]].mean())
```

##### Insight: "The more education the parents have, the higher the students score on average."



#### Question: Does type of lunch affect the scores of students?

```python
print(df.groupby("lunch")[["math score", "reading score", "writing score"]].mean())
```

##### Insight: "Students with standard lunch score comparatively higher in all subjects."

#### Question: Which subject is easier to score higher?

```python
print(df["math score"].mean())
print(df["reading score"].mean())
print(df["writing score"].mean())
```

##### Insight: "Students are more likely to score higher on reading."

#### Question: How likely are female to complete test preparation course as compared to male?

```python
df.groupby(["gender", "test preparation course"]).size()
```

##### Insight: "Female are more likely to complete test preparation course."

#### Question: Which race / ethinicity is most likely to complete the test preparation course?

```python
print(df.groupby(["race/ethnicity","test preparation course"]).size())
```

##### Insight: "Group C is more likely to complete test preparation course."
