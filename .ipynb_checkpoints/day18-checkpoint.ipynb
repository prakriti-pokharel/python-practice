# DAY 18 — Missing Data Handling

## isna()
Checks every cell in the DataFrame — returns True if missing, False if not.

df.isna() → full DataFrame of True/False

df.isna().sum() → count of missing values per column

df.isna().any().any() → single True/False — does ANY null exist anywhere?

## dropna()
Removes rows with missing values.
    
df.dropna() → drops any row with at least one null (default)

df.dropna(how='all') → drops only rows where ALL values are null

## fillna()
Fills missing values instead of dropping them.
    
df["column"].fillna(0) → fill nulls with 0

df["column"].fillna(df["column"].mean()) → fill nulls with average of that column

## When to drop vs fill:
- Drop → when the row is too incomplete to be useful

- Fill with mean → when you want to keep the row and estimate the missing value

- Fill with 0 → when missing means zero in that context

## Why None becomes NaN/float:
Integers cannot hold null values in pandas.
    
When None is added to an integer column, pandas converts it to float automatically.
    
NaN (Not a Number) is how pandas represents missing numeric values.

```python
import pandas as pd
import numpy as np

df = pd.read_csv("StudentsPerformance.csv")
df.head()
```

```python
# Check how many missing values exist in each column
print(df.isna().sum())
```

```python
# Check if any value in the entire DataFrame is null
print(df.isna().any().any())
```

```python
# Creating a small DataFrame with missing values on purpose
data = {
    "name": ["Ram", "Sita", "Hari", "Gita"],
    "age": [20, None, 22, None],
    "score": [85, 90, None, 78]
}
df2 = pd.DataFrame(data)
```

```python
# Drop all rows that have any missing value
df2.dropna(how='any')
```

```python
# Fill the missing age with the average age
df2["age"] = df2["age"].fillna(df2["age"].mean())
print(df2)
```

```python
# Fill missing score with 0
df2["score"] = df2["score"].fillna(0)
print(df2)
```
