# DAY 21 — Introduction to Matplotlib

## Basic plot syntax
df.groupby("column")["value"].mean().plot(kind='bar')

## Always add these three lines to every chart:
plt.title("Your chart title")

plt.xlabel("X axis label")

plt.ylabel("Y axis label")

## plt.show()
Needed in .py terminal scripts to display the chart.
    
Not needed in Jupyter — charts display automatically.

## Chart types (kind=)
'bar'   → bar chart (categories vs values)

'line'  → line chart (trends over time)

'hist'  → histogram (distribution of one column)

'scatter' → scatter plot (relationship between two columns)

## Why visualize?
Same insight lands faster as a chart than as numbers.

Always visualize after groupby or filtering to communicate findings clearly.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("StudentsPerformance.csv")
```

#### Task: Make a bar chart showing the average math score by gender

```python
df.groupby("gender")["math score"].mean().plot(kind='bar')
plt.title("Average Math Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Math Score")
plt.show() #optional for Jupyter
```
