```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("StudentsPerformance.csv")
```

```python
# Print first 5 rows
print(df.head())
```

```python
# Filter students who scored above 70 in math
print(df[df['math score'] > 70])
```

```python
# Group by gender and get average math score
result = df.groupby('gender')['math score'].mean()
```

```python
# Create a bar chart of the groupby result
result.plot(kind = 'bar')
plt.title('Average math score by gender')
plt.ylabel('Average math score')
```
