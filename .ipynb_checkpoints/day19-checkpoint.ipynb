# DAY 19 — Merging DataFrames

## pd.merge()
Combines two DataFrames based on a common column.
    
syntax: pd.merge(df1, df2, on="common column", how="type")

## how='inner'
Keeps only rows that exist in BOTH DataFrames.
    
Non-matching rows are dropped entirely.

## how='outer'
Keeps ALL rows from both DataFrames.

Missing values appear as NaN where there is no match.

## Other merge types:
how='left' → keeps all rows from left DataFrame, matches where possible

how='right' → keeps all rows from right DataFrame, matches where possible

## Real world use:
You will almost always be merging two datasets that share a common column like ID, name, date, or region. Inner when you only want complete data, outer when you want to see everything including gaps.

```python
import pandas as pd

students = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "name": ["Ram", "Sita", "Hari", "Gita"]
})

scores = pd.DataFrame({
    "id": [1, 2, 3, 5],
    "math": [85, 90, 78, 92]
})
```

```python
# Merge keeping only students that exist in BOTH DataFrames
merged_df = pd.merge(students, scores, on='id', how='inner')
print(merged_df)
```

```python
# Merge keeping ALL students even if they have no score
merged_df = pd.merge(students, scores, on='id', how='outer')
print(merged_df)
```
