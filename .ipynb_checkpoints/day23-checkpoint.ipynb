# DAY 23 — Scatter Plot

## Scatter Plot
Shows the relationship between two numerical columns.
    
df.plot(kind='scatter', x='column1', y='column2')

## Reading a scatter plot:
Bottom-left to top-right diagonal → positive correlation

(high in one = high in the other)



Top-left to bottom-right diagonal → negative correlation 

(high in one = low in the other)



Random scattered dots → no correlation

(the two columns are unrelated)


```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("StudentsPerformance.csv")
```

```python
df.plot(kind='scatter', x='math score', y='reading score')
plt.title("Math Score vs Reading Score")
plt.xlabel("Math Score")
plt.ylabel("Reading Score")
plt.show()
```

## Finding from this dataset:
Math score and reading score show positive correlation.
    
Students who score high in math tend to score high in reading too.
    
The subjects are related — strong performance in one predicts strong performance in the other.

## Note: Correlation vs Causation:
Correlation means two things move together.
    
It does NOT mean one causes the other.
    
High math scores don't cause high reading scores —

both are likely influenced by a third factor like study habits or parental education.
