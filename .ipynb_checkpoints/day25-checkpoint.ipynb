## DAY 25 — Combining pandas operations

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("StudentsPerformance.csv")
df["average score"] = (df["math score"] + df["reading score"] + df["writing score"]) / 3
```

```python
# Step 1: Groupby to get average score by test preparation
# Store groupby result in a variable, not back into df

result = df.groupby("test preparation course")["average score"].mean()
print(result)

# Step 2 - plot it
result.plot(kind="bar")
plt.title("Average Score by Test Preparation")
plt.ylabel("Average Score")
plt.show()
```

```python
# Step 3: Create pass/fail column using lambda (from Day 15)
df["result"] = df["average score"].apply(lambda x: "Pass" if x > 40 else "Fail")

# Step 4: Group by two columns and count rows with .size()
# .size() counts rows per group without needing a column name
df.groupby(["test preparation course", "result"]).size()
```

### KEY REMINDER:
#### groupby result is a summary table — store in a variable

#### don't try to add it back as a df column

#### .size() for counting rows, .mean()/.sum() for calculations
