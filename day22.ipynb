# DAY 22 — Histogram

## Histogram
Shows the distribution of values in one column.
    
df["column"].plot(kind='hist')

## Normal Distribution
When a histogram peaks in the middle with fewer values on both sides.
    
Most students score near the average — very high and very low scores are rare.
    
This is called a normal distribution — a key concept in statistics.

## Key difference from bar chart:
Bar chart → compares categories (gender vs score)

Histogram → shows spread of one numerical column

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("StudentsPerformance.csv")
```

```python
df["math score"].plot(kind='hist')
plt.title("Distribution of Math Scores")
plt.show()
```
